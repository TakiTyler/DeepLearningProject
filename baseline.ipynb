# Baselines

The baseline we plan to have is **zero-shot prompting** with the base **Gemma 2B** model, where no fine-tuning has been applied. We will give it a list of ingredients and prompt it to output a recipe and measure the quality of the output. This helps establish what the model can do without any training.

````python
```python
import os
from huggingface_hub import login
from google.colab import userdata

# Securely retrieve the Hugging Face token from Colab Secrets
try:
    hf_token = userdata.get('HF_TOKEN')
    login(token=hf_token)
except Exception as e:
    print("Error: Please ensure you have added 'HF_TOKEN' to your Colab Secrets (🔑 sidebar).")

from transformers import AutoTokenizer, AutoModelForCausalLM
```
````

````python
# ============================================================
# Cell 1: Load the Pretrained Gemma-2B Tokenizer and Model
# ============================================================
# Load the tokenizer for Google's Gemma-2B model from Hugging Face Hub.
# The tokenizer handles converting raw text into token IDs and vice versa.
tokenizer = AutoTokenizer.from_pretrained("google/gemma-2b")
# Load the Gemma-2B causal language model with device_map="auto",
# which automatically distributes the model across available GPUs/CPU.
model = AutoModelForCausalLM.from_pretrained("google/gemma-2b", device_map="auto")
````

````python
# ============================================================
# Cell 2: Prepare the Input Prompt and Tokenize It
# ============================================================
# Define the natural-language prompt asking the model to generate
# a recipe from a list of ingredients (baseline test prompt).

input_text = "List of ingredients: Onion, Garlic, Salt, Pepper, Olive Oil, chicken. Output a recipe with at least these ingredients"
# Tokenize the prompt: convert the text into PyTorch tensors (return_tensors="pt")
# and move them to the GPU ("cuda") for accelerated inference.
input_ids = tokenizer(input_text, return_tensors="pt")
# Run inference: feed the tokenized input to the model's generate()
# method, which autoregressively predicts the next tokens to
# produce a recipe completion based on the prompt.

outputs = model.generate(**input_ids)
# Decode the generated token IDs back into human-readable text
# and print the model's recipe output. outputs[0] selects the
# first (and only) sequence from the batch.
print(tokenizer.decode(outputs[0]))
````
